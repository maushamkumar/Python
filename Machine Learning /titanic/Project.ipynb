## KNN Impute 

It's a way to hanle missing values by estimating the missing values based on the values of the nearest neighbors. It leverage the KNN Algorithm to find the most similar data point to the data point with missing values and then imputes the missing value on these neighbours. 

### How it works 
1. Step 1: Identify the data point with missing value 
2. Step 2: Use KNN Algorithm to find the "k" closet data points(neighbors) based on the feature that are not missing. The closeness is usually measured using a distance metric like Euclidean distance. 
3. Step 3: Impute the missing value using the values of the nearest neighbors. There are few ways to do this: 

For continuous variables: Use the mean or weighted average of the neighbors. 

For categorical variables: Use the mode of the nearest neighbors. 

```python
knn_data = data.copy()
```

```python
slicing_2=knn_data.iloc[:7,:4]
slicing_2
```

```python
slicing_2.drop(columns='Sex', inplace=True)
```

```python
from sklearn.impute import KNNImputer
knn_imputor = KNNImputer(n_neighbors=2)


# Impute missing value 
imputed_data = knn_imputor.fit_transform(slicing_2)
imputed_data
```

```python
# Convert this into DataFrame 
imputed_data1 = pd.DataFrame(imputed_data, columns=slicing_2.columns)
imputed_data1
```

## Iterative Impute 
Iterative imputer is a sophisticated method for handling missing values in dataset. It models each feature with missing values as a function of other features and uses an iterative process to fill the missing data. This approach is particularly effective for dataset where the relationship between features can be leverged for more accurate imputation. 

### Key Features of Iterative Imputer. 

- **Model-Based Imputation**
The iterative Impute fits a regression model for each feature with missing values using the other feature in the dataset as predictors. This allows for more nuanced approach compared to similar method like mean or median imputation. 

- **Iterative Process** 
The imputer works in an iterative manner: It starts with by filling in the missing values with an initial guess (like mean imputation) then refine these estimates by fitting the models multiple time until convergence is reached. 

- **Flexable with Different Models:** 
You can specify the estimator used for the imputation process. By default, It uses BayesianRidge regression, but you can use other regressors like Decision Tree, Random Forests, or any other scikit-learn estimator. Handling Multivariate Relationship: 

Since it consider the relationship between multiple features, it can produce more accurate imputations compared to univariate methods. 

### How it works 
- **Initialization:** Missing value are initially filled using a simple imputation method (like mean or median)
- **Model Fitting:**  For each feature with missing values, the imputer fit a regression model using using the other features. 
- **Imputation:** The model predict missing values for that feature. 
- **Itertion:** Step 2 and 3 are repeated for a specified number of iterations or until the predictions stabilize (convergence)

#### Pros and Cons: 
**Pros:** 
More accurate imputations because it takes into account the correlations between features. Can handle different types of features(numerical and categorical)

**Cons:** 
Computationally more intensive compared to simple imputation methods, especially or large dataset. Requires more tuning and understanding of the underlying relationship in the data.
