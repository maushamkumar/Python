# Encoding

## What Encoding Does
- Encoding converts categorical features into numerical features by applying various transformations.

## Why Do We Need Encoding?
- We need encoding because machine learning algorithms cannot work with categorical features directly.
- Algorithms (computers) only understand numbers.

## What Are the Ways to Encode Features?
- There are several prebuilt encoders available, and you can also build your own custom encoder.

---

## Types of Categorical Data
There are two types of categorical data:

1. **Ordinal Data**
   - Categories that have an inherent order.
   - Example: `Poor`, `Average`, `Good`
   - Order: `Good > Average > Poor`

2. **Nominal Data**
   - Categories that have no inherent order.
   - Example: `Pen`, `Paper`, `Notebook`
   - No category can be ranked as greater or smaller.

---

## Ordinal Categorical Encoding
- To convert ordinal categorical data into numerical data, scikit-learn provides **Ordinal Encoding**.

## How Ordinal Encoding Works
```bash
OrdinalEncoder(categories=[['Poor', 'Average', 'Good']])
```
Resulting Mapping

Poor → 1

Average → 2

Good → 3

## Label Encoding

- We use **Label Encoding** on the **output (target) column**.
- `LabelEncoder` does **not follow any mathematical or logical order**.
- It assigns values **based on alphabetical order**.
- There is **no semantic meaning** behind the assigned numbers.

### Key Characteristics
- No logic
- No inherent order
- Purely alphabetical mapping

### Example
```text
Classes: Ham, Spam
```
After Label Encoding:

Ham  → 0

Spam → 1

Note: LabelEncoder starts from 0, not 1.

### When Should We Use LabelEncoder?

- We **must use LabelEncoder for classification problems** to encode the target variable.

This works because class labels are **identifiers**, not quantities.

The model does **not assume any mathematical relationship** between classes.

### Why Don’t We Use Label Encoding on Input Features?
#### Reason:

- LabelEncoder does **not preserve any meaningful order.**

It assigns values purely based on **alphabetical order.**

As a result, the model may learn **false relationships** between categories.

It fails to capture any **real-world meaning** of the feature.

Example (Wrong Usage)
```text
Education Level: High School, Bachelor, Master
```
Alphabetical encoding:
```text
Bachelor    → 0
High School → 1
Master      → 2
``` 
This order is logically incorrect and misleads the model.

### Important Note
- You can technically apply LabelEncoder to input features,

- but it is **not recommended** because it introduces **artificial order** and **incorrect assumptions** into the model.

## LabelBinarizer

````python
from sklearn.preprocessing import LabelBinarizer

# Sample target variable for a multi-class classification problem 
y = ['cat', 'dog', 'fish', 'dog', 'cat']

# Initialize the LabelBinarizer 
lb = LabelBinarizer()

# Fit and transform the target variable
y_binarized = lb.fit_transform(y)

print("Binarized labels:\n", y_binarized)

# Inverse transform to recover original labels
y_original = lb.inverse_transform(y_binarized)

print("Original labels:\n", y_original)
````
