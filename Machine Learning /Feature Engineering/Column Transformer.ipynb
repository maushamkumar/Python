### The Hard way

```python
df['owner'].value_counts()
```

```python
# Apply ordinal encoder to owner 
oe = OrdinalEncoder(categories=[['Test Drive Car', 'Fourth & Above Owner', 'Third Owner', 'Second Owner', 'First Owner']])

X_train_owner = oe.fit_transform(X_train.loc[:, ['owner']])
X_test_owner = oe.transform(X_test.loc[:, ['owner']])
```

```python
# Convert to df 
X_train_owner_df = pd.DataFrame(X_train_owner, columns=oe.get_feature_names_out())
X_test_owner_df = pd.DataFrame(X_test_owner, columns=oe.get_feature_names_out())
```

### The Easy way

Let's implement this in easy way. 

```python
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('cars.csv')
```

```python
df.head()
```

Here we are also going to encode column in same way 
- Owner - We'll apply OrdinalEncoding 
- brand & fuel - we'll apply OneHotEncoding

```python
# Split the data into training set and testing set 
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['selling_price']), 
                                                    df['selling_price'], 
                                                    test_size=0.2,
                                                    random_state=42)
```

```python
transformer = ColumnTransformer(
    [
        ("ordinal", OrdinalEncoder(
                categories=[[
                    'Test Drive Car',
                    'Fourth & Above Owner',
                    'Third Owner',
                    'Second Owner',
                    'First Owner'
                ]],
                handle_unknown='use_encoded_value',
                unknown_value=-1
            ), ['owner']),
        
        ("onehot", OneHotEncoder(sparse_output=False, handle_unknown='ignore'), ['brand', 'fuel'])
    ],
    remainder='passthrough'
)

# setting to get a pandas df
# transformer.set_output(transform='pandas')
```

```python
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)
```
