```python
import pandas as pd
import os

def excel_to_csv(excel_file, output_csv):
    try:
        # Read Excel file
        df = pd.read_excel(excel_file)
        
        # Handle any null values
        df = df.fillna('')  # Replace NaN with empty string
        
        # Clean the data - remove non-ASCII characters or replace them
        for column in df.columns:
            if df[column].dtype == 'object':  # Only process string/object columns
                df[column] = df[column].str.encode('ascii', errors='ignore').str.decode('ascii')
        
        # Clean column names
        df.columns = df.columns.str.strip().str.replace(' ', '_')
        df.columns = df.columns.str.replace('[^0-9a-zA-Z_]', '', regex=True)
        
        # Convert to CSV with proper encoding and quoting
        df.to_csv(output_csv, 
                 index=False, 
                 encoding='ascii',
                 quoting=1,
                 escapechar='\\',
                 errors='ignore')
        
        print(f"Successfully converted {excel_file} to {output_csv}")
        
    except Exception as e:
        print(f"Error converting file: {str(e)}")

# Example usage
excel_file = "Sample - Superstore.xls"
output_csv = "output.csv"

excel_to_csv(excel_file, output_csv)
```
