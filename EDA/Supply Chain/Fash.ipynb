## Product Performance 
- Product performance tell you hwo well each product (or SKU) is selling. It's basically about **demand** and **sales volume**

**Why we look for this:**
- To identify **best-sellers** (SKUs that drive business).
- To identify **low-performance products** (which might need discounting, marketing push, or even removal)
- To track **inventory alignment**(We are stocking the right products?).

**Beyond basic analysis, you can also check:**
- **Seasonality:** Do some products sell more in certain months/seasons?
- **Profitability:** Not just which products sell the most, but which bring the most **Profit**(high-margin SKUs)
- **Trend analysis:** Are some products gaining or losing popularity over time? 

```python
data['Number of products sold']
```

```python
sku_sales = data.groupby('SKU')['Number of products sold'].sum().reset_index()
sku_sales = sku_sales.sort_values(by='Number of products sold', ascending=False)
```
