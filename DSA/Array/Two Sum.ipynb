🧩 Problem Statement: Two Sum
🔍 Description:
Given an array of integers nums and an integer target, return the indices of the two numbers such that they add up to the target.

You may assume that each input will have exactly one solution, and you may not use the same element twice.

You can return the answer in any order


🧪 Example:
Input:
nums = [2, 7, 11, 15], target = 9
Output:
[0, 1]
Explanation:
Because nums[0] + nums[1] == 9, we return [0, 1].

🚦Constraints:
2 <= nums.length <= 10⁴

-10⁹ <= nums[i] <= 10⁹

-10⁹ <= target <= 10⁹

Only one valid answer exists.

```python
nums = [2, 7, 11, 15]
target = 9

num_map = {}  # Correct dictionary

for i, num in enumerate(nums):  # Correct loop
    complement = target - num

    if complement in num_map:
        print(num_map[complement], i)
        break  # Optional: exit after finding the solution

    num_map[num] = i

else:
    print("No two sum solution found")  # Optional: only runs if loop doesn't break
```

```python
def two_sum(num, target):
    left = 0
    right = len(num) - 1
    if len(num) < 2:
        return None
    num = sorted(num)
    
    while left < right:
        current_sum = num[left] + num[right]
        
        if current_sum == target:
            return "Yes"

        elif current_sum < target:
            left += 1
        elif current_sum > target:
            right -= 1
```

```python
nums = [2, 7, 11, 15]
target = 12
print(two_sum(nums, target)) 
```
