```python
arr = [1, 0,7, 2, 0, 0, 3, 0, 4]
```

```python
temp = []
for i in range(len(arr)):
    if arr[i] != 0:
        temp.append(arr[i])
for i in range(len(arr)):
    if arr[i] == 0:
        temp.append(arr[i])
```

```python
print(temp)  # Output the modified array with zeros moved to the end
```

```python
# First step is find the first zero element 
j = -1 
for i in range(len(arr)):
    if arr[i] == 0:
        j = i 
        break
if j != -1:  # If a zero was found
    print(f"First zero found at index {arr}")
    
for i in range(j + 1, len(arr)):
    if arr[i] != 0:
        arr[j], arr[i] = arr[i], arr[j]
        j += 1
print(arr)  # Output the modified array with zeros moved to the end
```
