```python
nums = [2, 5, 6, 8, 11]
temp = nums[0]
n = len(nums)
for i in range(1, n):
    
    nums[i - 1] = nums[i]

nums[n - 1] = temp
print(nums)
```

```python
d = 3 
nums = [2, 5, 6, 8, 11]
temp = nums[:d]
print(temp)
n = len(nums)
for i in range(d, n):
    nums[i - d] = nums[i]
for i in range(d):
    nums[n - d + i] = temp[i]
print(nums)
```
