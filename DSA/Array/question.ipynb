## Find the largest element in an Array 


```python
arr = [3, 2, 1, 5, 2]
max(arr)
```

```python
max = arr[0]
for i in range(len(arr) - 1):
    if arr[i] < arr[i+1]:
        max = arr[i+1]
        
print(max)
```

## Find the second largest element

```python
arr = [1, 2, 4, 7, 7, 5]
largest = arr[0]
slargest = -1

for i in range(len(arr) - 1):
    if arr[i] < arr[i+1]:
        largest = arr[i+1]
        
    if slargest <= arr[i+1] and largest > arr[i+1]:
        slargest = arr[i+1]
        
print(slargest)
```

## Remove dublicate from a sorted array

```python
arr = [1, 1, 2, 2, 2, 3, 3]
```

```python
arr
```

```python
for i in range(len(arr) - 1):
    for j in range(len(arr)-1, i, -1):
        if arr[j] == arr[i]:
            del arr[j]
            
print(arr)
```

```python
arr = [1, 2, 2,2, 2, 2, 2, 3, 4]
```

```python
for i in range(len(arr)):
    j = i + 1 
    while j < len(arr):
        if arr[i] == arr[j]:
            del arr[j]
        else:
            j += 1
print(arr)
```

```python
remove_duplicate = set(arr)
print(list(remove_duplicate))
```

##### Modify the array and return the number of unique array. 

```python
arr = [1, 1, 2, 2, 2, 3, 3]
temp = []
for i in range(len(arr)-1):
    if arr[i] == arr[i+1]:
        continue
    else:
        temp.append(arr[i])
print(len(temp))
```
