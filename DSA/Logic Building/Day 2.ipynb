## Task:
1. Ignore all numbers that are negative or zero.
2. From the remaining numbers, keep only those that have exactly 3 digits.
3. From those numbers, keep only the ones whose middle digit is odd.
4. From the filtered list, return the sum of the first and last digit of the largest such number.
5. If nothing survives, return None.

```python
def is_three_digit(nums): 
    if len(str(nums)) == 3: 
        return True
    else: 
        return False
   
def is_middle_digit_odd(nums):
    if int(str(nums)[1]) % 2 != 0: 
        return True
    
def sum_of_first_and_last_number(nums): 
    return int(str(nums)[0]) + int(str(nums)[-1])

def task(nums):
    maximum = None
    for i in nums:
        
        if i < 0: 
            continue
        
        if not is_three_digit(i): 
            continue
        
        if not is_middle_digit_odd(i):
            continue
        
        if maximum is None or i > maximum: 
            maximum = i
            
    final_sum = sum_of_first_and_last_number(maximum)
    
    return final_sum

nums = [105, 212, 303, 480, -123, 907, 111, 89]
task(nums)
```

# Actual Answer 

```python
def is_three_digit(n):
    return 100 <= n <= 999

def is_middle_digit_odd(n):
    return int(str(n)[1]) % 2 != 0

def sum_first_last(n):
    s = str(n)
    return int(s[0]) + int(s[-1])

def task(nums):
    maximum = None

    for n in nums:
        if n <= 0:
            continue
        
        if not is_three_digit(n):
            continue
        
        if not is_middle_digit_odd(n):
            continue
        
        if maximum is None or n > maximum:
            maximum = n

    if maximum is None:
        return None

    return sum_first_last(maximum)

nums = [105, 212, 303, 480, -123, 907, 111, 89]
print(task(nums))
```

## Task:
1. Ignore all numbers that are negative or zero.
2. From the remaining numbers, keep only those that have exactly 4 digits.
3. From those numbers, keep only the ones whose product of digits is divisible by 6.
4. From the filtered list, return the second smallest number.
5. If fewer than 2 numbers survive, return None.

nums = [1024, 1111, 2301, -4000, 3605, 999, 1206, 8400]


```python
nums = [1024, 1111, 2301, -4000, 3605, 999, 1206, 8400]
```

```python
def is_four_digit(n):
    return 1000 <= n <= 9999

def is_divisible(nums):
    
    product = 1
    
    for n in str(nums):
        product = product * int(n)
    
    return product % 6 == 0

maximum = None
for n in nums:
    if n <= 0:
        continue
    
    if not is_four_digit(n):
        continue
    
    if not is_divisible(n):
        continue
    
    if maximum is None or n > maximum:
        maximum = n
    
    
print(maximum)
```

```python
nums = [1024, 1111, 2301, -4000, 3605, 999, 1206, 8400]

def is_four_digit(n):
    return 1000 <= n <= 9999

def product_divisible_by_6(n):
    product = 1
    for ch in str(n):
        product *= int(ch)
    return product % 6 == 0

def find_second_smallest(nums):
    smallest = None
    second_smallest = None

    for n in nums:
        if n <= 0:
            continue
        
        if not is_four_digit(n):
            continue
        
        if not product_divisible_by_6(n):
            continue
        
        if smallest is None or n < smallest:
            second_smallest = smallest
            smallest = n
        
        elif n > smallest and (second_smallest is None or n < second_smallest):
            second_smallest = n

    return second_smallest

print(find_second_smallest(nums))
```
