## Task
1. You start with balance = 0
2. Traverse the list from left to right
3. At each step:
- Add the number to balance
- If balance ever becomes negative, reset it to 0

4. Track the maximum balance you ever reached during the process
5. Return that maximum balance


This is not max number, not sum of all positives — it’s about running state.

```python
nums = [5, -3, 4, -2, -1, 6, -7, 8]
```

```python
balance = 0 
max = None 
for i in nums: 
    balance += i
    if max is None or balance > max: 
        max = balance
    
print(balance)
```

```python
balance = 0 
max_balance = 0 

for n in nums: 
    
    balance += n 
    
    if balance < 0: 
        balance = 0 
        
    if balance > max_balance:
        max_balance = balance
print(max_balance)
```

## Task

Find the length of the longest consecutive failure streak.

Rules:

- Traverse once, left to right.

- Maintain state.

- No counters reset lazily — you must understand when and why.

```python
events = [1, 0, 0, 1, 0, 0, 0, 1, 0, 0]

max_failure = 0
failure = 0
for i in events:
    
    if i == 1:
        failure = 0
    else:  
        if i == 0:
            failure += 1 
            
            if failure > max_failure:
                max_failure = failure
print(max_failure)
            
        
    
```

```python
events = [1, 0, 0, 1, 0, 0, 0, 1, 0, 0]

current_strike = 0 
max_strike = 0 

for event in events: 
    
    if event == 0: 
        current_strike += 1 
        if current_strike > max_strike: 
            max_strike = current_strike
            
    else: 
        current_strike = 0 
print(max_strike)
```

## Task

Count how many clicks are accepted.

Rules:

- The first click is always accepted.

- After a click is accepted, ignore all clicks that happen within the next 5 seconds.

- A click is accepted only if:

current_click_time - last_accepted_time >= 5


```python
clicks = [1, 3, 5, 6, 12, 13, 20]

Accepted_click = 0 
current_click = 0
for i in clicks:
    if clicks[0] == i: 
        Accepted_click += 1 
        current_click = i
    
    else: 
        if (i - current_click) >= 5:
            Accepted_click += 1 
            current_click = i
print(Accepted_click) 
    
```

```python
clicks = [1, 3, 5, 6, 12, 13, 20]

accepted_count = 0
last_accepted = None

for t in clicks:
    if last_accepted is None:
        accepted_count += 1
        last_accepted = t
    elif t - last_accepted >= 5:
        accepted_count += 1
        last_accepted = t

print(accepted_count)
```
