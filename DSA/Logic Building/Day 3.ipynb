```python
nums = [123321, 456654, 120021, 998899, 111111, 543345, -678876, 987789]

def is_six_digit(nums): 
    return 100000 <= nums <= 999999

def valid_digit_sum(nums): 
    s = str(nums)
    
    return int(s[0])+int(s[1])+int(s[2]) == int(s[-1])+int(s[-2])+int(s[-3])

def find_second_smallest(nums):
    smallest = None
    second_smallest = None 

    for n in nums: 
        
        if not is_six_digit(n): 
            continue
        
        if not valid_digit_sum(n):
            continue
        
        if smallest is None or n < smallest:
            second_smallest = smallest
            smallest = n
        elif n > smallest and (second_smallest is None or n < second_smallest): 
            second_smallest = n 
            
    return second_smallest

print(find_second_smallest(nums))
```
