## 🧠 Problem (logic-heavy, beginner-safe)

You’re given a list of integers.


Task:

1. Ignore all negative numbers
2. From the remaining numbers:
    -  keep only even numbers
    - square them

3. Return the sum of those squared values

```python
nums = [3, -2, 4, 7, -5, 10]
sum = 0 
for i in range(len(nums)):
    if nums[i] > 0 and nums[i] % 2 == 0: 
        sum = sum + nums[i] ** 2
print(sum)
```

```python
def process_numbers(nums):
    total = 0

    for n in nums:
        if n < 0:
            continue
        
        if n % 2 == 0:
            total += n * n

    return total


nums = [3, -2, 4, 7, -5, 10]
print(process_numbers(nums))
```

```python
nums = [121, 44, 10, -11, 7, 131, 22]

def is_palindrome(n):
    s = str(n)
    return s == s[::-1]

def digit_sum_is_prime(n):
    total = 0
    for ch in str(n):
        total += int(ch)

    if total <= 1:
        return False

    i = 2
    while i * i <= total:
        if total % i == 0:
            return False
        i += 1

    return True

def find_smallest_valid(nums):
    minimum = None

    for n in nums:
        if n < 0:
            continue

        if not is_palindrome(n):
            continue

        if not digit_sum_is_prime(n):
            continue

        if minimum is None or n < minimum:
            minimum = n

    return minimum

print(find_smallest_valid(nums))
```

```python
def is_prime(n):
    if n <= 1:
        return False

    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1

    return True


n = 3

if is_prime(n):
    print("Prime")
else:
    print("Not Prime")
```
