#### Recursion

When a function calls itself until a specifice condition is met.

```python
def fun():
    print(1)
    fun()
    
```

```python
fun()
```

```python
cnt = 0 
def fun2():
    global cnt 
    cnt += 1
    if cnt < 5:
        print(cnt)
        fun2()
```

```python
fun2()
```

```python
cnt = 0 
def call(name):
    global cnt 
    if cnt < 5:
        return
    print(name)
    cnt += 1 
    call(name)
        
```

```python
call("Mausham")
```

#### Print linearly 1 to N using Recusrion

```python
cnt = 1
def f1(n):
    global cnt 
    if cnt == n: 
        return
    print(cnt)
    cnt += 1
    f1(n)
```

```python
f1(10)
```

```python
def f2(n, cnt = 1):
    if cnt > n:
        return
    print(cnt)
    f2(n, cnt + 1)
    
    
```

```python
f2(10)
```

```python
def f3(n, current, index = 0):
    if index == n:
        return
    print(current)
    f3(n, current - 1, index + 1)
    
```

```python
f3(5, 5)
```

### Sum of first N Number

```python

def f1(n):
    sum = 0
    if n == 0:
        return 0
    sum += n + f1(n - 1)
    # print(sum)
    return sum
    
    
```

```python
f1(5)
```

```python

def f1(n):
    mul = 1
    if n == 0:
        return 1
    mul *= n * f1(n - 1)
    # print(sum)
    return mul
    
    
```

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

```python
factorial(5)
```

#### How to reverse an array 


```python
arr = [1, 2, 3, 4, 5]
out = [5, 4, 3, 2, 1]
```

```python
def reverse_array(arr, left=0, right=None):
    if right is None:
        right = len(arr) - 1
    if left >= right:
        return
    arr[left], arr[right] = arr[right], arr[left]
    reverse_array(arr, left + 1, right - 1)
```

```python
arr = [1, 2, 3, 4, 5]
reverse_array(arr)
print(arr)  # [5, 4, 3, 2, 1]
```

```python
def reverse_array(arr, index=0):
    if index >= len(arr) // 2:
        return
    arr[index], arr[len(arr) - 1 - index] = arr[len(arr) - 1 - index], arr[index]
    reverse_array(arr, index + 1)
```

```python
def is_palindrome(arr, index=0):
    if index >= len(arr) // 2:
        return "Palindrome"
    if arr[index] != arr[len(arr) - 1 - index]:
        return "Not Palindrome"
    return is_palindrome(arr, index + 1)
```

```python
print(is_palindrome([1, 2, 3, 2, 1]))  
print(is_palindrome([1, 2, 3, 4, 5]))  
```
